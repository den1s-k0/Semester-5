```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_excel('irisdat.xlsx')
print(data.shape)
data.head(3)
```

```python
x = data[data['type'] == 3]['sepalwid'].values
print(x.shape)
x[:10]
```

```python
#для случайной выборки данные должны быть из одного распределения и должны быть независимы
```

```python
x.mean().item()
```

```python
x.var().item()
```

```python
x.std().item()
```

```python
std = x.std().item()
mean = x.mean().item()
le = mean - 3*std
ri = mean + 3*std
print(f'({le:.3f}; {ri:.3f})')
```

```python
x[(x < le) | (x > ri)]
```

```python
#используется правило трех сигм, что данные не отличаются более чем на +/- 3*std
#значений не найдено -- значит удовлетворяет, иначе удаляем неподходящие(только для больших объемов данных)
```

```python
x.min(), x.max()
```

```python
x.mean().item(), np.quantile(x, 0.5).item()
```

```python
#среднее арифм - неустойчивое, медиана - устойчивая(центральный элемент отсортированной выборки)
```

```python
np.quantile(x, 0.05).item(), np.quantile(x, 0.95).item()
```

```python
np.quantile(x, 0.75).item() - np.quantile(x, 0.25).item()
```
