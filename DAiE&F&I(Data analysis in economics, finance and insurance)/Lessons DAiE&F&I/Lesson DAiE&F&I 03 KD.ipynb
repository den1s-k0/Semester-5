```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as sts
```

```python
np.random.seed(27)
x = np.random.rand(100)
```

```python
plt.hist(x, bins = 7)
```

```python
# sts.chisquare
freq, edges = np.histogram(x, bins=4)
reference_distr = sts.Normal()
reference_distr.cdf(0)
```

```python
reference_distr.cdf(edges)
```

```python
edges_inf = edges.copy()
edges_inf[0] = -np.inf
edges_inf[-1] = np.inf
edges_inf
```

```python
p = reference_distr.cdf(edges_inf[1:]) - reference_distr.cdf(edges_inf[:-1])
np_ = p * len(x)
np_
```

```python
#теоретические вероятности
```

```python
freq
```

```python
freq.sum(), np_.sum()
```

```python
sts.chisquare(freq, np_, ddof = 1)
```
